## 1. Unit of analysis + time window

he unit of analysis is a single web page (represented by content_id). Each row represents the search performance metrics and traffic trends for one specific piece of content over a fixed snapshot period.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
url = 'https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

print(f"Total rows: {len(df):,}")
print(f"Unique content IDs: {df['content_id'].nunique():,}")
print("Conclusion: Since Total Rows equals Unique content IDs, the grain is exactly one row per content_id.")
```

Context: content_id and client_id (used strictly for identification and joining, never for training).

## 3. Verify it with queries (grain, counts, missing values, windows)
I will verify these claims by checking for missing values in our key features, and I will analyze the distribution of our proxy label (trend_pct) to see if a -15% drop is a reasonable threshold for our dataset.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("--- Missing Values in Key Features ---")
print(df[['search_volume', 'ctr', 'avg_position', 'trend_pct']].isnull().sum())

print("\n--- Label Distribution (trend_pct) ---")
print(df['trend_pct'].describe())

decay_count = len(df[df['trend_pct'] <= -15.0])
print(f"\nPages with a drop of 15% or worse: {decay_count:,} out of {len(df):,} ({decay_count/len(df):.1%})")
```
