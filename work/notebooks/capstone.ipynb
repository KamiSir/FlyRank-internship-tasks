Title: Predicting Content Decay: A Machine Learning Approach to Prioritizing SEO Refreshes

Abstract
Content teams frequently waste hours updating pages that do not need it, while true traffic decay goes unnoticed. To solve this, we framed content decay as a binary classification task to predict which pages are actively losing organic traffic based on historical search metrics. Using a Random Forest model on an anonymized 30,000-row dataset, we achieved a precision of 0.61 (validated via a rigorous GroupKFold split), significantly outperforming baseline heuristics. This directional, decision-support score allows content teams to efficiently prioritize thin, aging content for immediate editorial review.

## 1. Question

Introduction & Problem Statement
The core business decision is determining which specific URLs should be prioritized for a content update this week. A false positive wastes a writer's time (4-5 hours per page), while a false negative results in permanent traffic loss. Relying on fixed heuristic rules is too brittle. Machine learning can weigh multiple competing factors simultaneously to distinguish true content decay from normal seasonal variance.

## 2. Data

This analysis uses the content_refresh_anonymized.csv dataset, representing a snapshot of ~30,000 web pages.

The Unit of Analysis: One row represents one unique URL (content_id).

Excluded Data: Qualitative text/HTML was excluded, as were unique identifiers (client_id, content_id) to prevent model memorization. Post-window metrics (e.g., clicks_90d) and direct derivatives of the label (trend_pct) were strictly excluded to prevent target leakage.

## 3. Methodology
We framed this as a binary classification task.

Label: Our target (is_decaying) is a proxy label defined as a traffic drop worse than -15% (trend_pct <= -15.0).

Features: search_volume, word_count, and content_age_days.

Validation Design: To prevent the model from overfitting to a specific client's niche, we used a GroupKFold split grouped by client_id. This ensures the model is evaluated honestly on unseen clients.


## 4. Results (vs baseline)

Our hypothesis that older, thinner content decays faster was confirmed during our signal audit.

Naive Random Split Precision: 0.63

Honest Grouped Split Precision: 0.61
The honest split proves the model learned generalized decay patterns rather than client-specific quirks. It provides a reliable ~3x lift in precision compared to random guessing or basic heuristic baselines.

## 5. Limitations

This model provides directional, decision-support scores based on historically observed patterns. It does not claim absolute causal proof of why traffic dropped, nor can it predict exact future rankings. It is not valid for brand-new pages (under 90 days old) that lack a historical baseline.

## 6. Ranked recommendations

The model outputs a probability score predicting decay risk. Pages scoring above 0.60 are flagged and paired with human-readable reason codes (e.g., "Aging content + Thin word count"). Editors must manually review these flagged pages to ensure topic relevance before assigning a rewrite.
